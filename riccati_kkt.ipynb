# QP Resolution through Riccati recursion
We consider the following structured optimal control problem:

$$
\begin{align}
\min\limits_{\mathbf{x},\mathbf{u}} & \frac{1}{2} \sum_{k=0}^{N-1} \left( \begin{bmatrix}\mathbf{x}_k^T & \mathbf{u}_k^T \end{bmatrix}\begin{bmatrix}\mathbf{Q}_k & \mathbf{S}^T_k\\\mathbf{S}_k & \mathbf{R}_k \end{bmatrix} \begin{bmatrix}\mathbf{x}_k \\ \mathbf{u}_k \end{bmatrix} + \begin{bmatrix}\mathbf{q}_k^T & \mathbf{r}_k^T \end{bmatrix}\begin{bmatrix}\mathbf{x}_k \\ \mathbf{u}_k \end{bmatrix}\right) + \frac{1}{2}\mathbf{x}_N^T \mathbf{P}_N\mathbf{x}_N + \mathbf{p}_N^T\mathbf{x}_N \\ 
s.t. & \quad \mathbf{x}_{k+1} = \mathbf{A}_k\mathbf{x}_{k} + \mathbf{B}_k\mathbf{u}_{k} + \mathbf{b}_k
\end{align}
$$

We start considering the case $N = 2$, hence $\mathbf{x} = \begin{bmatrix}\mathbf{x}_0 & \mathbf{x}_1\end{bmatrix}^T$ and $\mathbf{u} = \begin{bmatrix}\mathbf{u}_0\end{bmatrix}$. In this case the problem reduces to:

$$
\begin{align}
\min\limits_{\mathbf{x},\mathbf{u}} & \frac{1}{2} \begin{bmatrix}\mathbf{x}_0^T & \mathbf{u}_0^T \end{bmatrix}\begin{bmatrix}\mathbf{Q}_0 & \mathbf{S}^T_0\\\mathbf{S}_0 & \mathbf{R}_0 \end{bmatrix} \begin{bmatrix}\mathbf{x}_0 \\ \mathbf{u}_0 \end{bmatrix} + \begin{bmatrix}\mathbf{q}_0^T & \mathbf{r}_0^T \end{bmatrix}\begin{bmatrix}\mathbf{x}_0 \\ \mathbf{u}_0 \end{bmatrix} + \frac{1}{2}\mathbf{x}_1^T \mathbf{P}_1\mathbf{x}_1 + \mathbf{p}_1^T\mathbf{x}_1 \\ 
s.t. & \quad \mathbf{x}_{1} = \mathbf{A}_0\mathbf{x}_{0} + \mathbf{B}_0\mathbf{u}_{0} + \mathbf{b}_0
\end{align}
$$
with $\mathbf{P}_1 = \mathbf{Q}_1$ and $\mathbf{p}_1 = \mathbf{q}_1$. 

The Lagrangian function of this problem has the following form:

$$\mathcal{L} = \frac{1}{2}\mathbf{x}_0^T\mathbf{Q}_0\mathbf{x}_0 + \frac{1}{2}\mathbf{u}_0^T\mathbf{R}_0\mathbf{u}_0 + \mathbf{x}_0^T\mathbf{S}_0^T\mathbf{u}_0 + \mathbf{q}_0^T\mathbf{x}_0+ \mathbf{r}_0^T\mathbf{u}_0 + \frac{1}{2}\mathbf{x}_1^T \mathbf{P}_1\mathbf{x}_1 + \mathbf{p}_1^T\mathbf{x}_1 + \boldsymbol{\pi}^T\left(\mathbf{A}_0\mathbf{x}_{0} + \mathbf{B}_0\mathbf{u}_{0} -\mathbf{I}\mathbf{x}_1 + \mathbf{b}_0 \right)$$
with $\mathbf{I}$ identity matrix.

We now take derivatives of $\mathcal{L}$ w.r.t. the variables to form the KKT system:

\begin{align}
\frac{\partial\mathcal{L}}{\partial\mathbf{x}_0}& = \mathbf{Q}_0\mathbf{x}_0 + \mathbf{S}_0^T\mathbf{u}_0 + \mathbf{q}_0 + \mathbf{A}_0^T\boldsymbol{\pi} = \mathbf{0} \\
\frac{\partial\mathcal{L}}{\partial\mathbf{u}_0}& = \mathbf{R}_0\mathbf{u}_0 + \mathbf{S}_0\mathbf{x}_0 + \mathbf{r}_0 + \mathbf{B}_0^T\boldsymbol{\pi} = \mathbf{0} \\
\frac{\partial\mathcal{L}}{\partial\mathbf{x}_1}& = \mathbf{P}_1\mathbf{x}_1 + \mathbf{p}_1 - \mathbf{I}\boldsymbol{\pi} = \mathbf{0}\\
\frac{\partial\mathcal{L}}{\partial\boldsymbol{\pi}}& = \mathbf{A}_0\mathbf{x}_{0} + \mathbf{B}_0\mathbf{u}_{0} -\mathbf{I}\mathbf{x}_1 + \mathbf{b}_0 = \mathbf{0} 
\end{align}

that in matrix form has the form:

$$
\begin{bmatrix}
\mathbf{Q}_0 & \mathbf{S}_0^T  & \mathbf{0} & \mathbf{A}_0^T\\
\mathbf{S}_0 & \mathbf{R}_0 &  \mathbf{0} & \mathbf{B}_0^T \\
\mathbf{0} & \mathbf{0} &  \mathbf{P}_1 & -\mathbf{I} \\
\mathbf{A}_0 & \mathbf{B}_0 &  -\mathbf{I} & \mathbf{0} 
\end{bmatrix}
\begin{bmatrix}
\mathbf{x}_0 \\ 
\mathbf{u}_0 \\
\mathbf{x}_1 \\
\boldsymbol{\pi}
\end{bmatrix} = 
\begin{bmatrix}
-\mathbf{q}_0 \\
-\mathbf{r}_0 \\
-\mathbf{p}_1 \\
-\mathbf{b}_0
\end{bmatrix}
$$

or in a compact notation:

$$
\begin{bmatrix}
\mathbf{H} & \mathbf{C}^T\\
\mathbf{C} & \mathbf{0} 
\end{bmatrix}
\begin{bmatrix}
\mathbf{x}_0 \\ 
\mathbf{u}_0 \\
\mathbf{x}_1 \\
\boldsymbol{\pi}
\end{bmatrix} = 
\begin{bmatrix}
-\mathbf{g} \\
-\mathbf{b}_0
\end{bmatrix}
$$

```python
# Importing necessary libraries
import numpy as np
import matplotlib.pyplot as plt
# Setting print to allow full array display
np.set_printoptions(linewidth=10000)
from scipy.optimize import minimize
import time
```

```python
ns = 4
nu = 2

Q = []
Q.append(1. * np.eye(ns))
Q.append(1. * np.eye(ns))

q = []
q.append(.1 * np.ones((ns,1)))
q.append(.1 * np.ones((ns,1)))

R = []
R.append(0.1 * np.eye(nu))

r = []
r.append(0.1 * np.ones((nu,1)))

S = []
S.append(0.2 * np.ones((nu, ns)))

A = []
A.append(np.eye(ns, ns))

B = []
B.append(np.ones((ns, nu)))

b = []
b.append(np.ones((ns,1)))
```

```python
def create_problem(ns, nu, Q, R, S, q, r, A, B, b):
    problem_size = len(Q) * ns + len(R) * nu
    H = np.zeros((problem_size, problem_size))
    g = np.zeros((problem_size, 1))
    C = np.zeros(((len(Q)-1)*ns, problem_size))
    c = np.zeros(((len(Q)-1)*ns, 1))

    for k in range(0, len(R)):
        H[k*(ns+nu):k*(ns+nu)+ns+nu, k*(ns+nu):k*(ns+nu)+ns+nu] = np.block([[Q[k], S[k].T],[S[k], R[k]]])
        g[k*(ns+nu):k*(ns+nu)+ns + nu] = np.block([[q[k]], [r[k]]])
        
        C[k*ns:k*ns+ns, k*(ns+nu):k*(ns+nu)+ns+nu+ns] = np.block([A[k], B[k], -np.eye(ns)])
        c[k*ns:k*ns+ns] = -b[k]
        

    N = len(Q)-1
    H[N*(ns+nu):N*(ns+nu)+ns+nu, N*(ns+nu):N*(ns+nu)+ns+nu] = Q[len(Q)-1]
    g[N*(ns+nu):] = q[len(Q)-1]

    return H, g, C, c
    

H,g, C, c = create_problem(ns, nu, Q, R, S, q, r, A, B, b)
print(f"H: \n {H}")
print(f"g: \n {g}")
print(f"C: \n {C}")
print(f"c: \n {c}")
```

The KKT system can be solved by inverting the KKT matrix:

$$\begin{bmatrix}
\mathbf{x}_0 \\ 
\mathbf{u}_0 \\
\mathbf{x}_1 \\
\boldsymbol{\pi}
\end{bmatrix} =  \begin{bmatrix}
\mathbf{H} & \mathbf{C}^T\\
\mathbf{C} & \mathbf{0} 
\end{bmatrix}^{-1}\begin{bmatrix}
-\mathbf{g} \\
-\mathbf{b}_0
\end{bmatrix}$$

```python
KKT = np.block([
    [H, C.T],
    [C, np.zeros((ns, ns))]
])

rhs = np.zeros((2*ns + nu + ns,1))
rhs[0:2*ns + nu] = -g
rhs[2*ns + nu:] = c

tic = time.time()
w = np.linalg.pinv(KKT)@rhs
toc = time.time()

x = []
u = []

for k in range(0, len(Q)-1):
    x.append(w[k*(ns+nu):k*(ns+nu)+ns])
    u.append(w[k*(ns+nu)+ns:k*(ns+nu)+ns+nu])
    
x.append(w[k*(ns+nu)+ns+nu:k*(ns+nu)+ns+nu+ns])

for k in range(0, len(x)):
    print(f"x[{k}]: {x[k].T}")
for k in range(0, len(u)):
    print(f"u[{k}]: {u[k].T}")

invert_sol_time = toc - tic
print(f"in {invert_sol_time:.3g} sec or in {(invert_sol_time)*1000:.3g} milliseconds")
```

We can compare this result against an out-of-the-box solver: 

```python
constraints = {
    'type': 'eq',
    'fun': lambda x: C @ x - c.flatten(),
    'jac': lambda x: C
}

def objective(x):
    return 0.5 * x @ H @ x + g.flatten() @ x

def objective_jac(x):
    return H @ x + g.flatten()

# Initial guess
x0 = np.zeros(H.shape[0])

# Solve
tic = time.time()
res = minimize(objective, x0, jac=objective_jac,
               constraints=constraints,
               method='trust-constr')
toc = time.time()

print("Optimal value:", res.fun)

x = []
u = []
for k in range(0, len(Q)-1):
    x.append(res.x[k*(ns+nu):k*(ns+nu)+ns])
    u.append(res.x[k*(ns+nu)+ns:k*(ns+nu)+ns+nu])
x.append(res.x[-ns:])

for k in range(0, len(x)):
    print(f"x[{k}]: {x[k].T}")
for k in range(0, len(u)):
    print(f"u[{k}]: {u[k].T}")
opt_sol_time = toc - tic
print(f"in {opt_sol_time:.3g} sec or in {(opt_sol_time)*1000:.3g} milliseconds")
```

## Riccati recursion computation
To compute the Riccati recursion, we can do the following steps:

1. From $\frac{\partial\mathcal{L}}{\partial\boldsymbol{\pi}} = \mathbf{A}_0\mathbf{x}_{0} + \mathbf{B}_0\mathbf{u}_{0} -\mathbf{I}\mathbf{x}_1 + \mathbf{b}_0 = \mathbf{0}$:

   $$\mathbf{x}_1 = \mathbf{A}_0\mathbf{x}_{0} + \mathbf{B}_0\mathbf{u}_{0} + \mathbf{b}_0$$

2. We substitute in $\frac{\partial\mathcal{L}}{\partial\mathbf{x}_1} = \mathbf{P}_1\mathbf{x}_1 + \mathbf{p}_1 - \mathbf{I}\boldsymbol{\pi} = \mathbf{0}$:

    $$ \mathbf{P}_1\left(\mathbf{A}_0\mathbf{x}_{0} + \mathbf{B}_0\mathbf{u}_{0} + \mathbf{b}_0 \right) + \mathbf{p}_1 - \mathbf{I}\boldsymbol{\pi} = \mathbf{0} $$

  -> $$ \mathbf{P}_1\mathbf{A}_0\mathbf{x}_{0} + \mathbf{P}_1\mathbf{B}_0\mathbf{u}_{0} + \mathbf{P}_1\mathbf{b}_0  + \mathbf{p}_1 - \mathbf{I}\boldsymbol{\pi} = \mathbf{0} $$

  hence:

  $$\boldsymbol{\pi} = \mathbf{P}_1\mathbf{A}_0\mathbf{x}_{0} + \mathbf{P}_1\mathbf{B}_0\mathbf{u}_{0} + \mathbf{P}_1\mathbf{b}_0  + \mathbf{p}_1$$

3. We substitute in $\frac{\partial\mathcal{L}}{\partial\mathbf{x}_0} = \mathbf{Q}_0\mathbf{x}_0 + \mathbf{S}_0^T\mathbf{u}_0 + \mathbf{q}_0 + \mathbf{A}_0^T\boldsymbol{\pi} = \mathbf{0}$:

   $$ \mathbf{Q}_0\mathbf{x}_0 + \mathbf{S}_0^T\mathbf{u}_0 + \mathbf{q}_0 + \mathbf{A}_0^T\left( \mathbf{P}_1\mathbf{A}_0\mathbf{x}_{0} + \mathbf{P}_1\mathbf{B}_0\mathbf{u}_{0} + \mathbf{P}_1\mathbf{b}_0  + \mathbf{p}_1 \right) = \mathbf{0} $$

   -> $$ \mathbf{Q}_0\mathbf{x}_0 + \mathbf{S}_0^T\mathbf{u}_0 + \mathbf{q}_0 + \mathbf{A}_0^T\mathbf{P}_1\mathbf{A}_0\mathbf{x}_{0} + \mathbf{A}_0^T\mathbf{P}_1\mathbf{B}_0\mathbf{u}_{0} + \mathbf{A}_0^T\mathbf{P}_1\mathbf{b}_0  + \mathbf{A}_0^T\mathbf{p}_1 = \mathbf{0} $$

   -> $$\left(\mathbf{Q}_0 + \mathbf{A}_0^T\mathbf{P}_1\mathbf{A}_0 \right)\mathbf{x}_0 + \left(\mathbf{S}_0^T + \mathbf{A}_0^T\mathbf{P}_1\mathbf{B}_0 \right)\mathbf{u}_0 + \left[\mathbf{q}_0 + \mathbf{A}_0^T\left(\mathbf{P}_1\mathbf{b}_0 +   \mathbf{p}_1\right) \right] = \mathbf{0}$$

   We renamed the following matrices and vectors:

   $$
    \begin{align}
       \overline{\mathbf{Q}}_0 & = \mathbf{Q}_0 + \mathbf{A}_0^T\mathbf{P}_1\mathbf{A}_0 \\
       \overline{\mathbf{S}}_0^T & = \mathbf{S}_0^T + \mathbf{A}_0^T\mathbf{P}_1\mathbf{B}_0 \\
       \overline{\mathbf{q}}_0 & = \mathbf{q}_0 + \mathbf{A}_0^T\left(\mathbf{P}_1\mathbf{b}_0 +   \mathbf{p}_1\right)    
    \end{align}
   $$
   to obtain:

   $$ \overline{\mathbf{Q}}_0\mathbf{x}_0 + \overline{\mathbf{S}}_0^T\mathbf{u}_0 + \overline{\mathbf{q}}_0 = \mathbf{0}  $$

4. We substitute the result of step 2. in $\frac{\partial\mathcal{L}}{\partial\mathbf{u}_0} = \mathbf{R}_0\mathbf{u}_0 + \mathbf{S}_0\mathbf{x}_0 + \mathbf{r}_0 + \mathbf{B}_0^T\boldsymbol{\pi} = \mathbf{0}$. Following the same steps, we arrive at the form:

$$\left(\mathbf{R}_0 + \mathbf{B}_0^T\mathbf{P}_1\mathbf{B}_0 \right)\mathbf{u}_0 + \left(\mathbf{S}_0 + \mathbf{B}_0^T\mathbf{P}_1\mathbf{A}_0 \right)\mathbf{x}_0 + \left[\mathbf{r}_0 + \mathbf{B}_0^T\left(\mathbf{P}_1\mathbf{b}_0 + \mathbf{p}_1 \right) \right] = \mathbf{0} $$

We renamed the following matrices and vectors:

   $$
    \begin{align}
       \overline{\mathbf{R}}_0 & = \mathbf{R}_0 + \mathbf{B}_0^T\mathbf{P}_1\mathbf{B}_0 \\
       \overline{\mathbf{S}}_0 & = \mathbf{S}_0 + \mathbf{B}_0^T\mathbf{P}_1\mathbf{A}_0 \\
       \overline{\mathbf{r}}_0 & = \mathbf{r}_0 + \mathbf{B}_0^T\left(\mathbf{P}_1\mathbf{b}_0 +   \mathbf{p}_1\right)    
    \end{align}
   $$
   to obtain:

   $$ \overline{\mathbf{S}}_0\mathbf{x}_0 + \overline{\mathbf{R}}_0^T\mathbf{u}_0 + \overline{\mathbf{r}}_0 = \mathbf{0}  $$


The results from steps 3. and 4. can be combined in matrix form:

$$
\begin{bmatrix}
\overline{\mathbf{Q}}_0 & \overline{\mathbf{S}}_0^T \\
\overline{\mathbf{S}}_0 & \overline{\mathbf{R}}_0
\end{bmatrix}
\begin{bmatrix}
\mathbf{x}_0 \\
\mathbf{u}_0
\end{bmatrix} = 
\begin{bmatrix}
-\overline{\mathbf{q}}_0 \\
-\overline{\mathbf{r}}_0
\end{bmatrix} 
$$

5. From the second row we can compute the control $\mathbf{u}_0$ in function of the state $\mathbf{x}_0$:

   $$\mathbf{u}_0 = -\overline{\mathbf{R}}_0^{-1}\overline{\mathbf{r}}_0 - \overline{\mathbf{R}}_0^{-1}\overline{\mathbf{S}}_0\mathbf{x}_0$$

  We renamed the matrix and the vector:
  $$
    \begin{align}
        \mathbf{K}_0 &= -\overline{\mathbf{R}}_0^{-1}\overline{\mathbf{S}}_0\\
        \mathbf{k}_0 &= -\overline{\mathbf{R}}_0^{-1}\overline{\mathbf{r}}_0
    \end{align}
  $$

  hence:

  $$ \mathbf{u}_0 = {\mathbf{K}}_0\mathbf{x}_0 + \mathbf{k}_0 $$

  If we substitute the expression $\mathbf{u}_0 = -\overline{\mathbf{R}}_0^{-1}\overline{\mathbf{r}}_0 - \overline{\mathbf{R}}_0^{-1}\overline{\mathbf{S}}_0\mathbf{x}_0$ in the first row we can compute the state $\mathbf{x}_0$:

  $$\overline{\mathbf{Q}}_0\mathbf{x}_0 + \overline{\mathbf{S}}_0^T \left(-\overline{\mathbf{R}}_0^{-1}\overline{\mathbf{r}}_0 - \overline{\mathbf{R}}_0^{-1}\overline{\mathbf{S}}_0\mathbf{x}_0 \right) = -\overline{\mathbf{q}}_0 $$

  -> $$\left(\overline{\mathbf{Q}}_0 - \overline{\mathbf{S}}_0^T\overline{\mathbf{R}}_0^{-1}\overline{\mathbf{S}}_0\right)\mathbf{x}_0 = -\overline{\mathbf{q}}_0 + \overline{\mathbf{R}}_0^{-1}\overline{\mathbf{r}}_0 $$ 

    We renamed the matrix and the vector:
  $$
    \begin{align}
        \overline{\mathbf{P}}_0 &= \overline{\mathbf{Q}}_0 - \overline{\mathbf{S}}_0^T\overline{\mathbf{R}}_0^{-1}\overline{\mathbf{S}}_0\\
        \overline{\mathbf{p}}_0 &= \overline{\mathbf{q}}_0 - \overline{\mathbf{R}}_0^{-1}\overline{\mathbf{r}}_0
    \end{align}
  $$

  hence:

  $$ \mathbf{x}_0 = -\overline{\mathbf{P}}_0^{-1}\overline{\mathbf{p}}_0 $$

## Recursion
At iteration $k$, we consider the triplet $\mathbf{x}_k$, $\mathbf{u}_k$, and $\mathbf{x}_{k+1}$.

  1. In the *backward pass* we compute $\overline{\mathbf{P}}_k$, $\overline{\mathbf{p}}_k$, $\mathbf{K}_k$, and $\mathbf{k}_k$, inputting $\overline{\mathbf{P}}_{k+1}$, $\overline{\mathbf{p}}_{k+1}$ computed at the previous iteration.

  2. The *backward pass* is initialized with $\mathbf{P}_N = \mathbf{Q}_f$ and $\mathbf{p}_N = \mathbf{q}_f$ and computes backard until node 0 all the matrices $\mathbf{K}$ and vectors $\mathbf{k}$.

  3. In the *forward pass*, starting from node 0 to node N, we compute the control $\mathbf{u}_k = \mathbf{K}_k\mathbf{x}_k + \mathbf{k}_k$ and the next state $\mathbf{x}_{k+1} = \mathbf{A}_k\mathbf{x}_{k} + \mathbf{B}_k\mathbf{u}_{k} + \mathbf{b}_k$. If not given, we compute the initial state as $\mathbf{x}_0 = -\overline{\mathbf{P}}_0^{-1}\overline{\mathbf{p}}_0$.

```python
def riccati_backward(Q, S, R, q, r, P, p, A, B, b):
    Q_ = Q + A.T@P@A
    R_ = R + B.T@P@B
    S_ = S + B.T@P@A

    l = P@b + p
    q_ = q + A.T@l
    r_ = r + B.T@l

    R_i = np.linalg.pinv(R_)
    P_ = Q_ - S_.T@R_i@S_
    p_ = q_ - S_.T@R_i@r_

    K = -R_i @ S_
    k = -R_i @ r_

    return P_, p_, K, k

def riccati_forward(x0, K, k, A, B, b):
    u = K@x0 + k
    x = A@x0 + B@u + b

    return x,u

def initial_state(P, p):
    return -np.linalg.pinv(P)@p
```

```python
P = Q[1]
p = q[1]
tic = time.time()

x = []
u = []

x.append(np.zeros((ns, 1)))
x.append(np.zeros((ns, 1)))
u.append(np.zeros((nu, 1)))

P, p, K, k = riccati_backward(Q[0], S[0], R[0], q[0], r[0], P, p, A[0], B[0], b[0])
x[0] = initial_state(P, p)
x[1], u[0] = riccati_forward(x[0], K, k, A[0], B[0], b[0])
toc = time.time()

for k in range(0, len(x)):
    print(f"x[{k}]: {x[k].T}")
for k in range(0, len(u)):
    print(f"u[{k}]: {u[k].T}")

riccati_sol_time = toc - tic
print(f"in {riccati_sol_time:.3g} sec or in {(riccati_sol_time)*1000:.3g} milliseconds")
```

## Given initial state $\mathbf{x}_0$

We want to include the initial state $\mathbf{x}_0 = \mathbf{x}_m$.

This corresponds to a constraint:

$$\begin{bmatrix}\mathbf{I} & \mathbf{0} \end{bmatrix}
\begin{bmatrix}\mathbf{x}_0 \\ \mathbf{u}_0 \\ \vdots \\ \mathbf{x}_{N} \end{bmatrix} = \mathbf{x}_m
$$

to be stacked in the matrix of the constraints $\mathbf{C}$.

```python
xm = np.random.rand(4, 1)
print(f"xm: \n {xm.T}")

E = np.zeros((ns, C.shape[1]))
E[0:ns, 0:ns] = np.eye(ns)

C = np.vstack((C,E))
print(f"C: \n {C}")

c = np.concatenate((c, xm))
print(c)
```

Solving the KKT system

```python
KKT = np.block([
    [H, C.T],
    [C, np.zeros((C.shape[0], C.shape[0]))]
])

#print(f"KKT: \n {KKT}")

rhs = np.zeros((g.shape[0] + c.shape[0] ,1))
rhs[0:g.shape[0]] = -g
rhs[g.shape[0]:] = c


#print(f"rhs: \n {rhs}")

tic = time.time()
w = np.linalg.pinv(KKT)@rhs
toc = time.time()

x = []
u = []
for k in range(0, len(Q)-1):
    x.append(w[k*(ns+nu):k*(ns+nu)+ns])
    u.append(w[k*(ns+nu)+ns:k*(ns+nu)+ns+nu])
x.append(w[-ns:])

for k in range(0, len(x)):
    print(f"x[{k}]: {x[k].T}")
for k in range(0, len(u)):
    print(f"u[{k}]: {u[k].T}")

invert_N_initial_sol_time = toc - tic
print(f"in {invert_N_initial_sol_time:.3g} sec or in {(invert_N_initial_sol_time)*1000:.3g} milliseconds")
```

Solving the QP

```python
constraints = {
    'type': 'eq',
    'fun': lambda x: C @ x - c.flatten(),
    'jac': lambda x: C
}

def objective(x):
    return 0.5 * x @ H @ x + g.flatten() @ x

def objective_jac(x):
    return H @ x + g.flatten()

# Initial guess
x0 = np.zeros(H.shape[0])

# Solve
tic = time.time()
res = minimize(objective, x0, jac=objective_jac,
               constraints=constraints,
               method='trust-constr')
toc = time.time()

print("Optimal value:", res.fun)

x = []
u = []
for k in range(0, len(Q)-1):
    x.append(res.x[k*(ns+nu):k*(ns+nu)+ns])
    u.append(res.x[k*(ns+nu)+ns:k*(ns+nu)+ns+nu])
x.append(res.x[-ns:])

for k in range(0, len(x)):
    print(f"x[{k}]: {x[k].T}")
for k in range(0, len(u)):
    print(f"u[{k}]: {u[k].T}")

opt_N_initial_sol_time = toc - tic
print(f"in {opt_N_initial_sol_time:.3g} sec or in {(opt_N_initial_sol_time)*1000:.3g} milliseconds")
```

Solving using Riccati

```python
from collections import deque

x = []
u = []
x.append(np.zeros((ns, 1)))
u.append(np.zeros((nu, 1)))
for i in range(0, N):
    x.append(np.zeros((ns, 1)))
    u.append(np.zeros((nu, 1)))

x.append(np.zeros((ns, 1)))


tic = time.time()
P = Q[len(Q)-1]
p = q[len(Q)-1]
K = deque()
d = deque()

tic = time.time()
for k in range(len(Q)-1,0, -1):
    P, p, Kk, dk = riccati_backward(Q[k-1], S[k-1], R[k-1], q[k-1], r[k-1], P, p, A[k-1], B[k-1], b[k-1])
    K.appendleft(Kk)
    d.appendleft(dk)

x[0] = xm
for k in range(1, len(Q)):
    x[k], u[k-1] = riccati_forward(x[k-1], K[k-1], d[k-1], A[k-1], B[k-1], b[k-1])

toc = time.time()

for k in range(0, len(x)):
    print(f"x[{k}]: {x[k].T}")
for k in range(0, len(u)):
    print(f"u[{k}]: {u[k].T}")

riccati_N_optimal_sol_time = toc - tic
print(f"in {riccati_N_optimal_sol_time:.3g} sec or in {(riccati_N_optimal_sol_time)*1000:.3g} milliseconds")
```
